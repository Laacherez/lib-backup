# Plotting MSD loglog against time lags with theory.

```python
data = np.loadtxt("msd_vs_tau.txt")

tau      = data[:, 0]
msd_x    = data[:, 1]
msd_y    = data[:, 2]
msd_z    = data[:, 3]
stderr_x = data[:, 4]
stderr_y = data[:, 5]
stderr_z = data[:, 6]

plt.figure()

plt.plot(tau, msd_x, label="MSD x")
plt.fill_between(
    tau,
    msd_x - stderr_x,
    msd_x + stderr_x,
    alpha=0.3
)

plt.plot(tau, msd_y, label="MSD y")
plt.fill_between(
    tau,
    msd_y - stderr_y,
    msd_y + stderr_y,
    alpha=0.3
)

plt.plot(tau, msd_z, label="MSD z")
plt.fill_between(
    tau,
    msd_z - stderr_z,
    msd_z + stderr_z,
    alpha=0.3
)

plt.xscale("log")
plt.yscale("log")
plt.xlabel(r"$\tau$")
plt.ylabel(r"$\langle \Delta x^2 \rangle$")
plt.legend()
plt.tight_layout()
plt.show()
```
